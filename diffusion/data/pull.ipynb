```python
# MNIST
# download from kaggle and pull train-images-idx3-ubyte
# then

import numpy as np
from PIL import Image
import struct

with open('train-images-idx3-ubyte', 'rb') as f:
    magic, num, rows, cols = struct.unpack('>IIII', f.read(16))
    images = np.fromfile(f, dtype=np.uint8).reshape(num, rows, cols)

for i, img in enumerate(images):
    Image.fromarray(img).save(f'mnist_{i:05d}.png')

# shit ok just move all the pngs to mnist_png4 then (parity with alot)

# Structure is like mnist_png4/mnist_xxxxx.png
```
